# Day 04: Functions for Reusable Data Tasks

**Dataset:** `Employee_Dataset.csv`

This notebook creates reusable Python functions for calculations, data transformation, and
simple statistical operations, using real values pulled directly from the Employee dataset.

## 1. Import Libraries and Load the Dataset

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("../datasets/Employee_Dataset.csv")

# Preview the data
df.head()
```

```python
# Basic info about the dataset
df.info()
```

**Pull two numeric columns into plain Python lists, to use with our functions below**

```python
ages = list(df["Age"])
experience = list(df["ExperienceInCurrentDomain"])
leave_status = list(df["LeaveOrNot"])

print(ages[:10])
print(experience[:10])
```

## 2. Function 1: Calculate the Mean

`def` defines a function. This one takes a **parameter** called `values` and **returns** the
calculated average, instead of just printing it.

```python
def calculate_mean(values):
    total = 0
    for value in values:
        total += value
    return total / len(values)


# Call the function with real data and store the returned value
average_age = calculate_mean(ages)
print("Average age:", average_age)
```

## 3. Function 2: Calculate the Maximum

```python
def calculate_maximum(values):
    highest = values[0]
    for value in values:
        if value > highest:
            highest = value
    return highest


oldest_age = calculate_maximum(ages)
print("Oldest age:", oldest_age)
```

## 4. Function 3: Calculate the Minimum

```python
def calculate_minimum(values):
    lowest = values[0]
    for value in values:
        if value < lowest:
            lowest = value
    return lowest


youngest_age = calculate_minimum(ages)
print("Youngest age:", youngest_age)
```

## 5. Function 4: Calculate a Percentage

This function takes **two parameters**, `part` and `whole`, showing how a function can
combine more than one input to produce a single returned result.

```python
def calculate_percentage(part, whole):
    return (part / whole) * 100


# How many employees eventually left (LeaveOrNot == 1)?
employees_who_left = 0
for status in leave_status:
    if status == 1:
        employees_who_left += 1

leave_percentage = calculate_percentage(employees_who_left, len(leave_status))
print(f"Employees who left: {employees_who_left} out of {len(leave_status)}")
print(f"Leave percentage: {leave_percentage:.2f}%")
```

## 6. Function 5: Calculate the Range

This function **reuses** the `calculate_maximum` and `calculate_minimum` functions we already
built, showing how functions can call other functions.

```python
def calculate_range(values):
    return calculate_maximum(values) - calculate_minimum(values)


age_range = calculate_range(ages)
experience_range = calculate_range(experience)

print("Age range:", age_range)
print("Experience range:", experience_range)
```

## 7. Function 6: Count Values Above a Threshold

This function uses a **default parameter value** (`threshold=30`). If the caller doesn't
provide a value for `threshold`, it automatically falls back to `30`.

```python
def count_above_threshold(values, threshold=30):
    count = 0
    for value in values:
        if value > threshold:
            count += 1
    return count


# Using the default threshold (30)
count_default = count_above_threshold(ages)

# Overriding the default with a keyword argument
count_custom = count_above_threshold(ages, threshold=40)

print(f"Employees older than 30 (default): {count_default}")
print(f"Employees older than 40 (custom):  {count_custom}")
```

## 8. Function 7: Build a Full Summary

This function combines several of the functions above and **returns a dictionary** of
results, showing how a function's return value can itself be a more complex data type.

```python
def summarize_column(values, column_name):
    summary = {
        "column": column_name,
        "mean": calculate_mean(values),
        "maximum": calculate_maximum(values),
        "minimum": calculate_minimum(values),
        "range": calculate_range(values)
    }
    return summary


age_summary = summarize_column(ages, "Age")
experience_summary = summarize_column(experience, "ExperienceInCurrentDomain")

print(age_summary)
print(experience_summary)
```

## 9. Parameters and Arguments

A **parameter** is the name listed in a function's definition (like `values` and `threshold`
above). An **argument** is the actual value you pass in when you call the function. The
examples below call the same function with different arguments to show how a single function
can be reused for different tasks.

```python
# Same function (calculate_mean), different arguments each time
mean_age = calculate_mean(ages)                      # argument: ages
mean_experience = calculate_mean(experience)         # argument: experience

print("Mean age:", mean_age)
print("Mean experience:", mean_experience)
```

```python
# Same function (count_above_threshold), called with positional and keyword arguments
by_position = count_above_threshold(experience, 5)          # positional arguments
by_keyword = count_above_threshold(values=experience, threshold=5)  # keyword arguments

print(by_position)
print(by_keyword)
```

## Outcome

This notebook built seven reusable Python functions, such as `calculate_mean`, `calculate_maximum`,
`calculate_minimum`, `calculate_percentage`, `calculate_range`, `count_above_threshold`, and
`summarize_column`, using real values from the `Employee_Dataset.csv` file. Each function
takes one or more parameters and **returns** its result instead of only printing it, which let
the returned values be reused in further calculations (like combining `calculate_maximum` and
`calculate_minimum` inside `calculate_range`). I also practiced calling the same function with
