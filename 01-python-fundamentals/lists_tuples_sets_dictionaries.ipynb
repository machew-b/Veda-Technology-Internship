# Day 05: Python Lists, Tuples, Sets and Dictionaries

**Dataset:** `Employee_Dataset.csv`

This notebook works with Python's four core collection data structures, such as lists, tuples, sets,
and dictionaries, performing operations such as adding, removing, searching, and updating
elements, using real values pulled directly from the Employee dataset.

## 1. Import Libraries and Load the Dataset

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("../datasets/Employee_Dataset.csv")

# Preview the data
df.head()
```

```python
# Basic info about the dataset
df.info()
```

## 2. Lists

A list is an **ordered**, **mutable** (changeable) collection that allows duplicate values.
Written with square brackets `[]`.

**Create a list from a real column**

```python
cities = list(df["City"])
ages = list(df["Age"])

print(cities[:10])
print(ages[:10])
print(type(cities))
```

**Program demonstrating common list methods: `append()`, `remove()`, `sort()`, and more**

```python
# Start with a small working copy so we don't modify the full dataset
sample_ages = ages[:10]
print("Original list:", sample_ages)

# append() - add an item to the end
sample_ages.append(99)
print("After append(99):", sample_ages)

# remove() - remove the first matching value
sample_ages.remove(99)
print("After remove(99):", sample_ages)

# sort() - sort the list in place
sample_ages.sort()
print("After sort():", sample_ages)

# insert() - insert a value at a specific position
sample_ages.insert(0, 18)
print("After insert(0, 18):", sample_ages)

# pop() - remove and return the last item (or a specific index)
removed_value = sample_ages.pop()
print("Popped value:", removed_value)
print("After pop():", sample_ages)

# count() - count how many times a value appears
print("Count of 28 in full ages list:", ages.count(28))

# index() - find the position of a value
print("Position of first 28:", ages.index(28))

# len() - number of items
print("Number of items:", len(sample_ages))

# Slicing - grab a sub-section of the list
print("First 3 items:", sample_ages[:3])
```

## 3. Tuples

A tuple is an **ordered**, **immutable** (unchangeable) collection that also allows duplicate
values. Written with parentheses `()`. Once created, its contents cannot be added to, removed,
or changed.

**Create a tuple from real data**

```python
# Pair each employee's Education and City together as a tuple
employee_record = (df["Education"].iloc[0], df["City"].iloc[0], df["Age"].iloc[0])

print(employee_record)
print(type(employee_record))
```

**Program demonstrating common tuple methods**

```python
# Build a tuple of the first 10 ages
ages_tuple = tuple(ages[:10])
print("Tuple:", ages_tuple)

# count() - count occurrences of a value
print("Count of 28:", ages_tuple.count(28))

# index() - find the position of a value
if 28 in ages_tuple:
    print("Position of 28:", ages_tuple.index(28))

# Indexing and slicing work just like lists
print("First item:", ages_tuple[0])
print("First 3 items:", ages_tuple[:3])

# Tuples are immutable - you cannot append, remove, or sort them directly.
# To "change" a tuple, convert it to a list, modify it, then convert it back:
ages_list_from_tuple = list(ages_tuple)
ages_list_from_tuple.append(100)
ages_tuple_updated = tuple(ages_list_from_tuple)
print("Updated tuple (via list conversion):", ages_tuple_updated)
```

## 4. Sets

A set is an **unordered** collection of **unique** items, where duplicates are automatically
removed. Written with curly braces `{}`. Sets are useful whenever you need to find unique
values or compare groups of values.

**Create sets from real data**

```python
unique_cities = set(df["City"])
unique_education = set(df["Education"])

print(unique_cities)
print(unique_education)
```

**Program demonstrating common set methods**

```python
payment_tier_1 = set(df[df["PaymentTier"] == 1]["City"])
payment_tier_3 = set(df[df["PaymentTier"] == 3]["City"])

print("Cities with Payment Tier 1:", payment_tier_1)
print("Cities with Payment Tier 3:", payment_tier_3)

# union() - combine both sets, keeping only unique values
print("Union:", payment_tier_1.union(payment_tier_3))

# intersection() - values that appear in both sets
print("Intersection:", payment_tier_1.intersection(payment_tier_3))

# difference() - values in the first set but not the second
print("Difference:", payment_tier_1.difference(payment_tier_3))

# add() - add a new item
sample_set = set(["Bangalore", "Pune"])
sample_set.add("New Delhi")
print("After add():", sample_set)

# remove() - remove an item (raises an error if not found)
sample_set.remove("Pune")
print("After remove():", sample_set)

# discard() - remove an item safely (no error if not found)
sample_set.discard("Mumbai")  # "Mumbai" isn't in the set, but no error occurs
print("After discard() of a missing item:", sample_set)

# issubset() - check if one set's values are all contained in another
small_set = {"Bangalore"}
print("Is small_set a subset of unique_cities?", small_set.issubset(unique_cities))
```

## 5. Dictionaries

A dictionary is an **unordered** (insertion-ordered since Python 3.7+) collection of
`key: value` pairs. Keys must be unique, and each key is used to access its value. Written
with curly braces `{}`.

**Create a dictionary from a real employee record**

```python
employee = df.iloc[0]

employee_dict = {
    "Education": employee["Education"],
    "City": employee["City"],
    "Age": employee["Age"],
    "PaymentTier": employee["PaymentTier"]
}

print(employee_dict)
print(type(employee_dict))
```

**Program demonstrating common dictionary methods**

```python
# Access a value using its key
print("City:", employee_dict["City"])

# get() - access a value safely, with a default if the key doesn't exist
print("Department (default if missing):", employee_dict.get("Department", "Not specified"))

# keys(), values(), items()
print("Keys:", employee_dict.keys())
print("Values:", employee_dict.values())
print("Items:", employee_dict.items())

# Adding a new key-value pair
employee_dict["Gender"] = employee["Gender"]
print("After adding Gender:", employee_dict)

# update() - add or update multiple key-value pairs at once
employee_dict.update({"EverBenched": employee["EverBenched"], "LeaveOrNot": employee["LeaveOrNot"]})
print("After update():", employee_dict)

# pop() - remove a key and return its value
removed_value = employee_dict.pop("LeaveOrNot")
print("Removed value:", removed_value)
print("After pop():", employee_dict)

# 'in' - check whether a key exists
print("Is 'City' a key?", "City" in employee_dict)
print("Is 'Salary' a key?", "Salary" in employee_dict)
```

## 6. Comparison Table: Lists vs Tuples vs Sets vs Dictionaries

| Feature | List | Tuple | Set | Dictionary |
|---|---|---|---|---|
| Syntax | `[1, 2, 3]` | `(1, 2, 3)` | `{1, 2, 3}` | `{"key": "value"}` |
| Ordered | Yes | Yes | No | Yes (insertion order, Python 3.7+) |
| Mutable (changeable) | Yes | No | Yes | Yes |
| Allows duplicates | Yes | Yes | No | Keys must be unique (values can repeat) |
| Access by | Index (position) | Index (position) | No indexing — membership only | Key |
| Common use case | An ordered collection you'll modify (add/remove/sort) | A fixed collection that shouldn't change | Getting unique values or comparing groups | Looking up values by a meaningful label |
