# Day 12: NumPy Aggregations and Statistics

**Dataset:** `Iris_Dataset.csv`

This notebook uses NumPy functions to calculate sum, mean, median, minimum, maximum, standard
deviation, and variance, using real values pulled directly from the Iris dataset.

## 1. Import Libraries and Load the Dataset

```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("../datasets/Iris_Dataset.csv")

# Preview the data
df.head()
```

**Create a 1D array from one column, and a 2D array from all four numeric feature columns**

```python
petal_length = np.array(df["petal length (cm)"])

feature_columns = ["sepal length (cm)", "sepal width (cm)", "petal length (cm)", "petal width (cm)"]
features = np.array(df[feature_columns])

print("1D array shape:", petal_length.shape)
print("2D array shape:", features.shape)
```

## 2. Examples of Aggregation Functions

Each of the functions below reduces an array down to a single summary number.

**1. `sum()` — total of all values**

```python
print(petal_length.sum())
```

**2. `mean()` — average value**

```python
print(petal_length.mean())
```

**3. `np.median()` — the middle value when sorted**

```python
print(np.median(petal_length))
```

**4. `min()` — smallest value**

```python
print(petal_length.min())
```

**5. `max()` — largest value**

```python
print(petal_length.max())
```

**6. `std()` — standard deviation (how spread out the values are)**

```python
print(petal_length.std())
```

**7. `var()` — variance (the standard deviation, squared)**

```python
print(petal_length.var())
```

**8. `np.ptp()` — the range (max minus min), and `np.percentile()` — a specific percentile**

```python
print("Range (ptp):", np.ptp(petal_length))
print("25th percentile:", np.percentile(petal_length, 25))
print("75th percentile:", np.percentile(petal_length, 75))
```

## 3. Experimenting with `axis=0` and `axis=1`

On the 2D `features` array (150 rows × 4 columns), `axis=0` aggregates **down each column**
(one result per feature), while `axis=1` aggregates **across each row** (one result per
flower).

**`axis=0` — mean of each column (one value per feature, across all 150 flowers)**

```python
column_means = features.mean(axis=0)
print("Column means [sepal length, sepal width, petal length, petal width]:")
print(column_means)
```

**`axis=1` — mean of each row (one value per flower, across its 4 features)**

```python
row_means = features.mean(axis=1)
print("First 10 row means (average of that flower's 4 measurements):")
print(row_means[:10])
```

**Comparing the shapes returned by each axis**

```python
print("features.shape:", features.shape)
print("mean(axis=0).shape:", features.mean(axis=0).shape)   # one value per column -> shape (4,)
print("mean(axis=1).shape:", features.mean(axis=1).shape)   # one value per row -> shape (150,)
```

## 4. Statistical Analysis of the Dataset

A full statistical summary of all four numeric columns at once, using `axis=0`.

```python
summary = {
    "mean": features.mean(axis=0),
    "median": np.median(features, axis=0),
    "std": features.std(axis=0),
    "var": features.var(axis=0),
    "min": features.min(axis=0),
    "max": features.max(axis=0)
}

summary_df = pd.DataFrame(summary, index=feature_columns)
summary_df
```

## 5. Comparing Results Across Different Columns

```python
stds = features.std(axis=0)
for column_name, std_value in zip(feature_columns, stds):
    print(f"{column_name}: std = {std_value:.3f}")

most_variable_index = np.argmax(stds)
least_variable_index = np.argmin(stds)
print(f"\nMost variable feature: {feature_columns[most_variable_index]}")
print(f"Least variable feature: {feature_columns[least_variable_index]}")
```

## 6. Short Interpretation of the Results

Based on the statistics computed above:

- **Petal length** has by far the highest standard deviation (**≈1.76 cm**) and variance
  (**≈3.09**) of the four features, meaning petal length varies the most across the 150
  flowers in this dataset (ranges all the way from **1.0 cm to 6.9 cm**).
- **Sepal width** has the lowest standard deviation (**≈0.43 cm**) and variance (**≈0.19**),
  meaning it's the most consistent measurement across flowers, staying fairly close to its
  mean of **≈3.05 cm**.
- **Petal width** also has a fairly low mean (**≈1.20 cm**) but a moderate spread relative to
  that mean, suggesting, like petal length, it varies more between individual flowers than
  the sepal measurements do.
- Since the dataset contains three different species of iris, this spread in the petal
  measurements (compared to the more consistent sepal measurements) hints that petal length
  and petal width are likely more useful for **telling the three species apart** than the
  sepal measurements are.
