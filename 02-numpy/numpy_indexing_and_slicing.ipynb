# Day 10: NumPy Indexing and Slicing

**Dataset:** `Wine_Dataset.csv`

This notebook practices selecting individual elements, rows, columns, and ranges from NumPy
arrays using indexing and slicing, using real values pulled directly from the Wine dataset.

## 1. Import Libraries and Load the Dataset

```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("../datasets/Wine_Dataset.csv")

# Preview the data
df.head()
```

```python
# Basic info about the dataset
df.info()
```

**Create a 1D array from a single real column, and a 2D array from several real columns**

```python
alcohol = np.array(df["Alcohol"])

feature_columns = ["Alcohol", "Malic acid", "Ash", "Magnesium", "Color intensity"]
features = np.array(df[feature_columns])

print("1D array shape:", alcohol.shape)
print("2D array shape:", features.shape)
print(features[:5])
```

## 2. One-Dimensional Indexing

Python indexing is **zero-based**, where the first element is at position `0`, not `1`.

**Example 1: Access the first element (index 0)**

```python
print(alcohol[0])
```

**Example 2: Access the 5th element (index 4)**

```python
print(alcohol[4])
```

**Example 3: Access the last element using negative indexing**

```python
print(alcohol[-1])
```

**Example 4: Access the second-to-last element using negative indexing**

```python
print(alcohol[-2])
```

## 3. Two-Dimensional Indexing

For a 2D array, use `array[row, column]` to select a specific element.

**Example 1: Access a single element: row 0, column 0 (the first wine's Alcohol value)**

```python
print(features[0, 0])
```

**Example 2: Access a single element: row 2, column 3 (the 3rd wine's Magnesium value)**

```python
print(features[2, 3])
```

**Example 3: Access an entire row (all features for one wine)**

```python
print(features[0])
```

**Example 4: Access an entire column (all wines' Ash values)**

```python
ash_values = features[:, 2]
print(ash_values[:10])
```

**Example 5: Access an element using negative row and column indices**

```python
print(features[-1, -1])   # last row, last column: the last wine's Color intensity
```

## 4. Slicing

Slicing uses `start:stop` (and an optional `:step`) to select a **range** of values, instead
of a single element. The `stop` position is never included.

**Example 1: The first 5 alcohol values**

```python
print(alcohol[0:5])
```

**Example 2: Everything from index 5 onward**

```python
print(alcohol[5:10])   # showing just a small window (5 to 9) for a readable preview
```

**Example 3: Everything up to (but not including) index 5**

```python
print(alcohol[:5])
```

**Example 4: Every 2nd value, using a step**

```python
print(alcohol[0:10:2])
```

**Example 5: The last 5 alcohol values, using negative slicing**

```python
print(alcohol[-5:])
```

**Example 6: Slicing a 2D array: the first 3 rows, all columns**

```python
print(features[0:3, :])
```

**Example 7: Slicing a 2D array: all rows, the first 2 columns**

```python
print(features[:, 0:2][:5])   # showing the first 5 rows for a readable preview
```

**Example 8: Slicing a 2D array: a sub-block of rows and columns**

```python
print(features[0:3, 1:3])   # rows 0-2, columns 1-2
```
