# Day 13: NumPy Random Sampling

**Dataset:** Generated Random Dataset

This notebook generates random numbers and samples using NumPy, and explores reproducibility
using random seeds. Since this task is about generating data rather than working with an
existing file, the "dataset" here is a small simulated set of student data built with NumPy's
random functions.

## 1. Import Libraries

```python
import numpy as np
import pandas as pd
```

## 2. Random Integers

`np.random.randint()` generates random whole numbers within a given range.

**Example 1: A single random integer between 1 and 100**

```python
single_int = np.random.randint(1, 101)
print(single_int)
```

**Example 2: An array of 10 random integers between 1 and 100**

```python
int_array = np.random.randint(1, 101, size=10)
print(int_array)
```

**Example 3: A 2D array of random integers (simulating a small grid of values)**

```python
int_grid = np.random.randint(0, 10, size=(3, 4))
print(int_grid)
```

## 3. Random Floating-Point Numbers

**Example 1: `np.random.rand()`: random floats between 0 and 1**

```python
random_floats = np.random.rand(10)
print(random_floats)
```

**Example 2: `np.random.uniform()`: random floats within a custom range**

```python
random_uniform = np.random.uniform(50, 100, size=10)
print(random_uniform)
```

**Example 3: `np.random.randn()`: random floats from a normal (bell-curve) distribution**

```python
random_normal = np.random.randn(10)
print(random_normal)
```

## 4. Random Samples from an Array

`np.random.choice()` picks random elements out of an existing array.

**Example 1: Randomly sample 5 values from an array, without replacement**

```python
grades = np.array(["A", "B", "C", "D", "F"])
sample_no_replacement = np.random.choice(grades, size=5, replace=False)
print(sample_no_replacement)
```

**Example 2: Randomly sample 10 values from an array, with replacement (values can repeat)**

```python
sample_with_replacement = np.random.choice(grades, size=10, replace=True)
print(sample_with_replacement)
```

**Example 3: Randomly sample with custom probabilities for each value**

```python
# Make "A" far more likely to be picked than the other grades
weighted_sample = np.random.choice(grades, size=10, p=[0.5, 0.2, 0.15, 0.1, 0.05])
print(weighted_sample)
```

## 5. Reproducibility Using a Random Seed

By default, random numbers are different every time. Setting a **seed** with
`np.random.seed()` makes the "random" sequence repeatable.

**Without a seed: two runs produce different results**

```python
print(np.random.randint(1, 100, size=5))
print(np.random.randint(1, 100, size=5))
```

**With a seed: the exact same sequence is produced every time**

```python
np.random.seed(42)
print(np.random.randint(1, 100, size=5))

np.random.seed(42)   # reset to the same seed
print(np.random.randint(1, 100, size=5))   # identical output to the line above
```

**A different seed produces a different (but still reproducible) sequence**

```python
np.random.seed(7)
print(np.random.randint(1, 100, size=5))
```

## 6. Generating a Small Simulated Dataset

Using a fixed seed for reproducibility, generate a simulated dataset of 20 students, combining
random integers, random floats, and random sampling from an array of category labels.

```python
np.random.seed(101)

num_students = 20

student_ids = np.arange(1, num_students + 1)
study_hours = np.round(np.random.uniform(1, 10, size=num_students), 1)
scores = np.random.randint(40, 101, size=num_students)
grades = np.random.choice(["A", "B", "C", "D"], size=num_students, p=[0.25, 0.35, 0.25, 0.15])

simulated_df = pd.DataFrame({
    "student_id": student_ids,
    "study_hours": study_hours,
    "score": scores,
    "grade": grades
})

simulated_df
```

**Because a seed was set, re-running the generation produces the exact same dataset**

```python
np.random.seed(101)   # reset to the same seed used above

study_hours_repeat = np.round(np.random.uniform(1, 10, size=num_students), 1)
scores_repeat = np.random.randint(40, 101, size=num_students)

print("study_hours match:", np.array_equal(study_hours, study_hours_repeat))
print("scores match:", np.array_equal(scores, scores_repeat))
```

**A quick summary of the simulated dataset**

```python
simulated_df[["study_hours", "score"]].describe()
```
