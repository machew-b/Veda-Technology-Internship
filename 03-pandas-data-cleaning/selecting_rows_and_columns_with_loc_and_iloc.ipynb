# Day 16: Selecting Rows and Columns with loc and iloc

**Dataset:** `Titanic_Dataset.csv`

This notebook practices selecting rows and columns using Pandas `loc[]` and `iloc[]`, and
explores the difference between label-based and position-based selection, using real values
pulled directly from the Titanic dataset.

## 1. Import Libraries and Load the Dataset

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("../datasets/Titanic_Dataset.csv")

# Preview the data
df.head()
```

**A note on the index**

By default, `df`'s row labels are just integers 0, 1, 2, ..., the same as their positions.
To make the difference between `loc[]` (label-based) and `iloc[]` (position-based) obvious,
this notebook also creates a second version of the DataFrame indexed by `PassengerId`
instead, so the row *labels* and row *positions* are no longer the same numbers.

```python
df_by_id = df.set_index("PassengerId")
df_by_id.head()
```

## 2. `loc[]` Examples: Label-Based Selection

`loc[]` selects using **labels**, whether it is the actual index value or column name, not its numeric
position.

**Example 1: Select a single row by its label (default integer index)**

```python
print(df.loc[0])
```

**Example 2: Select a range of rows by label (both ends are included)**

```python
df.loc[0:4]
```

**Example 3: Select specific columns for every row**

```python
df.loc[:, ["Name", "Age", "Survived"]].head()
```

**Example 4: Select specific rows and specific columns together**

```python
df.loc[0:3, ["Name", "Pclass", "Fare"]]
```

**Example 5: Select a row using a real, non-default label (`PassengerId`)**

```python
# PassengerId 3 is Pclass 3 Passenger with different positions in the two DataFrames
print(df_by_id.loc[3])
```

**Example 6: Boolean filtering combined with `loc[]` to select rows and columns together**

```python
df.loc[df["Age"] > 70, ["Name", "Age", "Survived"]]
```

## 3. `iloc[]` Examples: Position-Based Selection

`iloc[]` selects using **integer positions**, counting from `0`, regardless of what the actual
row labels or column names are.

**Example 1: Select a single row by its position**

```python
print(df.iloc[0])
```

**Example 2: Select a range of rows by position (the end position is NOT included)**

```python
df.iloc[0:5]
```

**Example 3: Select specific columns by their position for every row**

```python
df.iloc[:, [3, 5, 1]].head()   # Name, Age, Survived columns, by position
```

**Example 4: Select specific rows and specific columns together, by position**

```python
df.iloc[0:4, [3, 2, 9]]   # first 4 rows; Name, Pclass, Fare columns
```

**Example 5: Select a row from `df_by_id` by position (NOT by its `PassengerId` label)**

```python
# This selects the 3rd row in the DataFrame's current order,
# whatever PassengerId that row happens to have.
print(df_by_id.iloc[2])
```

**Example 6: Select the last 3 rows and the first 2 columns using negative and positional indexing**

```python
df.iloc[-3:, 0:2]
```

## 4. Comparing `loc[]` and `iloc[]`

The clearest way to see the difference is to make the same request both ways on
`df_by_id`, where the row **labels** (`PassengerId`, starting at 1) and row **positions**
(starting at 0) are no longer the same numbers.

```python
# loc[3] looks for the row LABELED PassengerId == 3
print("loc[3] (label-based):")
print(df_by_id.loc[3][["Name", "Age"]])

print()

# iloc[3] looks for the row at POSITION 3 (the 4th row, PassengerId == 4)
print("iloc[3] (position-based):")
print(df_by_id.iloc[3][["Name", "Age"]])
```

**Another key difference: slice endpoints**

`loc[]` slices are **inclusive** of the end label. `iloc[]` slices are **exclusive** of the
end position, just like regular Python list slicing.

```python
print("loc[0:4] includes rows 0, 1, 2, 3, AND 4:")
print(len(df.loc[0:4]), "rows")

print()

print("iloc[0:4] includes rows 0, 1, 2, 3 only (position 4 is excluded):")
print(len(df.iloc[0:4]), "rows")
```

| | `loc[]` | `iloc[]` |
|---|---|---|
| Selects by | Label (index value / column name) | Integer position (starting at 0) |
| Slice `0:4` includes | Everything through label `4` (inclusive) | Positions 0, 1, 2, 3 only (exclusive of 4) |
| Works if index isn't 0, 1, 2... | Yes, it uses the actual labels | Yes, the position is unaffected by the labels |
| Column selection | By column name | By column position |
