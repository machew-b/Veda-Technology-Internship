# Day 01: Python Variables and Data Types

**Dataset:** `Student_Performance_Dataset.csv`

This notebook practices Python variables and the fundamental data types, such as integers,
floats, strings, booleans, lists, tuples, sets, and dictionaries, using real values pulled
directly from the Student Performance dataset, such as scores, grades, and demographic
information.

## 1. Import Libraries and Load the Dataset

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("../datasets/Student_Performance_Dataset.csv")

# Preview the data
df.head()
```

```python
# Basic info about the dataset
df.info()
```

**Pick one student record to use in the examples below**

```python
student = df.iloc[1]  # std-02
print(student)
```

## 2. Integers (`int`)

Whole numbers, with no decimal point.

```python
# The student's math score column loads in as text (str), so we convert it to an integer
math_score = int(student["math_score"])

print(math_score)
print(type(math_score))
```

## 3. Floats (`float`)

Numbers that include a decimal point.

```python
# Scores that are already stored as floats in the dataset
reading_score = student["reading_score"]
writing_score = student["writing_score"]
science_score = student["science_score"]

print(reading_score)
print(writing_score)
print(science_score)
print(type(reading_score))
print(type(writing_score))
print(type(science_score))
```

## 4. Strings (`str`)

Text data, written inside quotes.

```python
# Real values pulled from the dataset
roll_no = str(student["roll_no"])
gender = str(student["gender"])
race_ethnicity = str(student["race_ethnicity"])
grade = str(student["grade"])

print(roll_no)
print(gender)
print(race_ethnicity)
print(grade)
print(type(roll_no))
print(type(gender))
print(type(race_ethnicity))
print(type(grade))
```

## 5. Booleans (`bool`)

A data type that can only be `True` or `False`. In this dataset, `lunch` and
`test_preparation_course` are stored as `1`/`0`, which convert naturally into booleans.

```python
# Boolean values converted directly from the dataset's 0/1 columns
has_standard_lunch = bool(student["lunch"])
completed_test_prep = bool(student["test_preparation_course"])

print(has_standard_lunch)
print(completed_test_prep)
print(type(has_standard_lunch))
print(type(completed_test_prep))
```

## 6. Lists (`list`)

An ordered, changeable collection of items, written with square brackets `[]`.

```python
# All four of this student's subject scores, stored as a list
scores = [math_score, reading_score, writing_score, science_score]

print(scores)
print(type(scores))

# Note that lists are changeable (mutable)
scores.append(student["total_score"])
print(scores)
```

## 7. Tuples (`tuple`)

An ordered collection of items, written with parentheses `()`. Unlike lists, tuples are
**immutable**, i.e., they cannot be changed after they are created.

```python
# This student's roll number paired with their final grade
student_record = (roll_no, grade)

print(student_record)
print(type(student_record))

# Tuples support indexing, just like lists
print("Roll number:", student_record[0])
print("Grade:", student_record[1])
```

## 8. Sets (`set`)

An unordered collection of **unique** items, written with curly braces `{}`.

```python
# The unique grades that appear across the whole dataset
unique_grades = set(df["grade"])
print(unique_grades)
print(type(unique_grades))

# The unique race/ethnicity groups in the dataset
unique_groups = set(df["race_ethnicity"])
print(unique_groups)
```

## 9. Dictionaries (`dict`)

A collection of `key: value` pairs, written with curly braces `{}`.

```python
# Represent the student as a dictionary using their real values
student_dict = {
    "roll_no": roll_no,
    "gender": gender,
    "math_score": math_score,
    "reading_score": reading_score,
    "grade": grade
}

print(student_dict)
print(type(student_dict))

# Access a value using its key
print("Math score:", student_dict["math_score"])
```

## 10. Using `type()` to Identify Data Types

`type()` tells us exactly what data type a variable is — useful for checking data before
working with it.

```python
print(type(math_score))       # int
print(type(reading_score))    # float
print(type(gender))           # str
print(type(has_standard_lunch))  # bool
print(type(scores))           # list
print(type(student_record))   # tuple
print(type(unique_grades))    # set
print(type(student_dict))     # dict
```

## 11. Type Conversion

Python lets us convert one data type into another using functions like `int()`, `float()`,
`str()`, and `bool()`.

**Convert a string to a number** (this is we've done to `math_score` above)

```python
raw_math_score = student["math_score"]   # this comes in as a string
converted_math_score = int(raw_math_score)  # convert string -> int

print(raw_math_score, type(raw_math_score))
print(converted_math_score, type(converted_math_score))
```

**Convert an integer to a float**

```python
math_score_as_float = float(math_score)

print(math_score, type(math_score))
print(math_score_as_float, type(math_score_as_float))
```

**Convert a number to a string**

```python
total_score = student["total_score"]
total_score_text = str(total_score)

print(total_score, type(total_score))
print(total_score_text, type(total_score_text))
```

**Convert a float to an integer (drops the decimal part)**

```python
rounded_reading_score = int(reading_score)

print(reading_score, type(reading_score))
print(rounded_reading_score, type(rounded_reading_score))
```

**Convert a number to a boolean (0 is False, any other number is True)**

```python
print(bool(student["lunch"]))
print(bool(student["test_preparation_course"]))
```

## 12. Basic Operations on Each Data Type

**Arithmetic operations on numbers**

```python
total_score = math_score + reading_score + writing_score + science_score
average_score = total_score / 4
score_percentage = (total_score / 400) * 100

print("Total score:", total_score)
print("Average score:", average_score)
print("Score percentage:", score_percentage)
```

**String operations**

```python
summary_text = "Grade " + grade + " in " + race_ethnicity  # concatenation
grade_upper = grade.upper()                                   # method
group_length = len(race_ethnicity)                             # length

print(summary_text)
print(grade_upper)
print(group_length)
```

**List operations**

```python
print("Number of scores:", len(scores))
print("Highest score:", max(scores))
print("Lowest score:", min(scores))
print("Sum of scores:", sum(scores))
```

**Tuple operations**

```python
print("Tuple length:", len(student_record))
print("Is 'A' in the tuple:", "A" in student_record)
```

**Set operations**

```python
passing_grades = {"A", "B", "C"}
failing_grades = {"D", "F"}

all_grades = passing_grades.union(failing_grades)
print("All grades:", all_grades)
print("Is this student's grade a passing grade:", grade in passing_grades)
```

**Dictionary operations**

```python
print("Keys:", student_dict.keys())
print("Values:", student_dict.values())

# Add a new key-value pair
student_dict["total_score"] = total_score
print(student_dict)
```

## 13. Putting It All Together

Combining several data types in one summary, using only the real values pulled from the
dataset above.

```python
print(f"Student {roll_no} ({gender}, {race_ethnicity}) completed test prep: "
      f"{completed_test_prep}, scored {scores}, with a total of {total_score} "
      f"and a final grade of {grade}.")
```
