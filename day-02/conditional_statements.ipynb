**Dataset:** `Employee_Salary_Dataset.csv`

This notebook uses `if`, `elif`, and `else` statements to make decisions based on numerical data (`Experience_Years`, `Age`, `Salary`) and categorical data (`Gender`) from the employee salary dataset.

## 1. Import Libraries and Load the Dataset

## 2. Conditional Examples on a Single Employee

Below are conditional examples (`if` / `elif` / `else`) applied to a single record at a time.
We will use the **first employee** in the dataset to demonstrate each condition.

```python
# Grab the first employee as an example record
employee = df.iloc[0]
print(employee)
```

**Example 1: Check if the employee's experience is more than 5 years**

```python
if employee["Experience_Years"] > 5:
    print("This employee has more than 5 years of experience.")
else:
    print("This employee has 5 years of experience or less.")
```

**Example 2: Check if the employee's salary is above 500,000**

```python
if employee["Salary"] > 500000:
    print("High salary employee.")
else:
    print("Not a high salary employee.")
```

**Example 3: Check the employee's age group**

```python
if employee["Age"] < 25:
    print("Young employee (under 25).")
elif employee["Age"] < 40:
    print("Mid-career employee (25-39).")
else:
    print("Senior employee (40+).")
```

**Example 4: Gender the employee's sex**

```python
if employee["Gender"] == "Female":
    print("Employee is Female.")
elif employee["Gender"] == "Male":
    print("Employee is Male.")
else:
    print("Gender not specified.")
```

**Example 5: Experience-based job level**

```python
if employee["Experience_Years"] <= 2:
    print("Entry-level employee.")
elif employee["Experience_Years"] <= 10:
    print("Mid-level employee.")
else:
    print("Senior-level employee.")
```

**Example 6: Salary tier classification**

```python
if employee["Salary"] < 50000:
    print("Salary Tier: Low")
elif employee["Salary"] < 300000:
    print("Salary Tier: Medium")
elif employee["Salary"] < 1000000:
    print("Salary Tier: High")
else:
    print("Salary Tier: Very High")
```

**Example 7: Is the employee a minor (Age < 18)?**

```python
if employee["Age"] < 18:
    print("This employee is under 18.")
else:
    print("This employee is 18 or older.")
```

**Example 8: Salary per year of experience (avoid divide-by-zero)**

```python
if employee["Experience_Years"] == 0:
    print("Cannot compute salary per year of experience (0 years).")
else:
    ratio = employee["Salary"] / employee["Experience_Years"]
    print(f"Salary per year of experience: {ratio:,.2f}")
```

**Example 9: Retirement-age check**

```python
if employee["Age"] >= 60:
    print("Employee is at or near retirement age.")
else:
    print("Employee is not yet at retirement age.")
```

**Example 10: Check the employee's sex and salary range simultaneously**

```python
if employee["Gender"] == "Female" and employee["Salary"] > 200000:
    print("Female employee earning above 200,000.")
elif employee["Gender"] == "Male" and employee["Salary"] > 200000:
    print("Male employee earning above 200,000.")
else:
    print("Employee earning 200,000 or below.")
```

## 3. Examples Using Multiple Conditions (`and` / `or`)

These examples combine more than one condition together using `and` / `or` operators.

**Example 1: Check if the employee is highly experienced AND well paid**

```python
if employee["Experience_Years"] > 10 and employee["Salary"] > 500000:
    print("Highly experienced and well-paid employee.")
else:
    print("Does not meet both criteria (experience > 10 yrs and salary > 500,000).")
```

**Example 2: Check if the employee is young OR low level of experience (possible junior hire)**

```python
if employee["Age"] < 25 or employee["Experience_Years"] < 2:
    print("Likely a junior hire (young age or low experience).")
else:
    print("Not classified as a junior hire.")
```

**Example 3: Check if the employee is underpaid relative to experience (possible anomaly)**

```python
if employee["Experience_Years"] > 10 and employee["Salary"] < 100000:
    print("Possible anomaly: high experience but low salary.")
else:
    print("Salary appears consistent with experience level.")
```

**Example 4: Check if the employee is a female senior (for diversity reporting, for example)**

```python
if employee["Age"] >= 40 and employee["Gender"] == "Female":
    print("Senior female employee.")
elif employee["Age"] >= 40 and employee["Gender"] == "Male":
    print("Senior male employee.")
else:
    print("Not a senior employee.")
```

## 4. Classification Program Using Conditions

Here we classify a few individual employees using only `if` / `elif` / `else` statements
(salary tier and experience level). Each employee is checked one at a time using plain
conditional statements (for the purpose of this task, I did not incorporate loops or other functions that will simplify the coding practice).

**Classify Employee i (row index i-1)**

```python
row_index = 0 # classify the first employee
employee_1 = df.iloc[row_index]

if employee_1["Salary"] < 50000:
    salary_tier = "Low"
elif employee_1["Salary"] < 300000:
    salary_tier = "Medium"
elif employee_1["Salary"] < 1000000:
    salary_tier = "High"
else:
    salary_tier = "Very High"

if employee_1["Experience_Years"] <= 2:
    experience_level = "Entry-level"
elif employee_1["Experience_Years"] <= 10:
    experience_level = "Mid-level"
else:
    experience_level = "Senior-level"

print(f"Employee {int(employee_1['ID'])} -> Salary Tier: {salary_tier}, Experience Level: {experience_level}")
```

**Apply the classification program to the whole dataset**

```python
# Salary Tier for every employee
df["Salary_Tier"] = "Very High"                                                     
df.loc[df["Salary"] < 50000, "Salary_Tier"] = "Low"                                 
df.loc[(df["Salary"] >= 50000) & (df["Salary"] < 300000), "Salary_Tier"] = "Medium"  
df.loc[(df["Salary"] >= 300000) & (df["Salary"] < 1000000), "Salary_Tier"] = "High"  

# Experience Level for every employee
df["Experience_Level"] = "Senior-level"                                                        
df.loc[df["Experience_Years"] <= 2, "Experience_Level"] = "Entry-level"                        
df.loc[(df["Experience_Years"] > 2) & (df["Experience_Years"] <= 10), "Experience_Level"] = "Mid-level"  

df.head(10)
```

**Count how many employees fall into each Salary Tier**

```python
df["Salary_Tier"].value_counts()
```

**Count how many employees fall into each Experience Level**

```python
df["Experience_Level"].value_counts()
```

## 5. Simple Classification Program for One Employee

A small program that classifies one employee's experience, salary, and age using only
`if` / `elif` / `else` statements. This is useful as a quick decision-making tool.

```python
# Sample employee values to classify
experience_years = 12
age = 35
salary = 650000
gender = "Female"

print("     Employee Classification     ")

# Experience level
if experience_years <= 2:
    print("Experience Level: Entry-level")
elif experience_years <= 10:
    print("Experience Level: Mid-level")
else:
    print("Experience Level: Senior-level")

# Salary tier
if salary < 50000:
    print("Salary Tier: Low")
elif salary < 300000:
    print("Salary Tier: Medium")
elif salary < 1000000:
    print("Salary Tier: High")
else:
    print("Salary Tier: Very High")

# Age group
if age < 25:
    print("Age Group: Young")
elif age < 40:
    print("Age Group: Mid-career")
else:
    print("Age Group: Senior")

# Combined condition
if experience_years > 10 and salary > 500000:
    print("Note: Highly experienced and well-paid.")
elif experience_years > 10 and salary < 100000:
    print("Note: Possible pay anomaly (high experience, low salary).")

print(f"Gender: {gender}")
```
