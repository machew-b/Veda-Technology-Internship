# Day 03: Loops for Dataset Processing

**Dataset:** `Student_Marks_Dataset.csv`

This notebook practices `for` loops and `while` loops to process collections of values and
perform repetitive data operations, using real values pulled directly from the Student Marks
dataset.

## 1. Import Libraries and Load the Dataset

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("../datasets/Student_Marks_Dataset.csv")

# Preview the data
df.head()
```

```python
# Basic info about the dataset
df.info()
```

**Convert the `Marks` column into a plain Python list, to loop over below**

```python
marks = list(df["Marks"])
number_courses = list(df["number_courses"])

print(marks[:10])
print(len(marks))
```

## 2. `for` Loop Examples

**Example 1: Loop through the first 5 marks and print each one**

```python
for mark in marks[:5]:
    print(mark)
```

**Example 2: Loop with `range()` to print marks by position**

```python
for i in range(5):
    print(f"Student {i + 1}: {marks[i]} marks")
```

**Example 3: Loop with `range()` and a step, to check every 10th student**

```python
for i in range(0, len(marks), 10):
    print(f"Student {i + 1}: {marks[i]} marks")
```

**Example 4: Loop through two lists together using `zip()`**

```python
for courses, mark in zip(number_courses[:5], marks[:5]):
    print(f"Took {courses} courses -> scored {mark} marks")
```

**Example 5: Loop through a dictionary's key-value pairs**

```python
# A small dictionary built from real dataset values
student_record = {
    "number_courses": number_courses[0],
    "time_study": df["time_study"].iloc[0],
    "marks": marks[0]
}

for key, value in student_record.items():
    print(f"{key}: {value}")
```

## 3. `while` Loop Examples

**Example 1: Print marks one at a time using a `while` loop**

```python
index = 0
while index < 5:
    print(marks[index])
    index += 1
```

**Example 2: Use a `while` loop to find the first student who scored above 50**

```python
index = 0
while index < len(marks):
    if marks[index] > 50:
        print(f"First student above 50 marks is at position {index}, with {marks[index]} marks.")
        break
    index += 1
```

**Example 3: Countdown of how many students are left to process**

```python
remaining = 5
while remaining > 0:
    print(f"{remaining} students left to process...")
    remaining -= 1
print("All done!")
```

## 4. Program: Process a List of Numerical Values

Using a `for` loop, count how many students passed (marks >= 40) versus how many did not,
across the whole dataset.

```python
passed_count = 0
failed_count = 0

for mark in marks:
    if mark >= 40:
        passed_count += 1
    else:
        failed_count += 1

print("Passed:", passed_count)
print("Failed:", failed_count)
```

## 5. Program: Calculate Summary Statistics Using Loops

Instead of relying on built-in functions like `sum()`, `max()`, and `min()`, this program uses
**accumulator variables** inside a `for` loop to build up the total, then find the highest and
lowest marks, one student at a time.

```python
total_marks = 0
highest_mark = marks[0]
lowest_mark = marks[0]

for mark in marks:
    total_marks += mark          # accumulator for the running total

    if mark > highest_mark:
        highest_mark = mark      # keep track of the highest mark seen so far

    if mark < lowest_mark:
        lowest_mark = mark       # keep track of the lowest mark seen so far

average_marks = total_marks / len(marks)

print("Total marks:", total_marks)
print("Average marks:", average_marks)
print("Highest mark:", highest_mark)
print("Lowest mark:", lowest_mark)
```

## 6. Looping Through a Dictionary: Counting `number_courses`

Building and looping through a dictionary is a common data-processing pattern. Here we count
how many students took each number of courses, using a dictionary as a running tally.

```python
course_counts = {}

for courses in number_courses:
    if courses in course_counts:
        course_counts[courses] += 1
    else:
        course_counts[courses] = 1

print(course_counts)
```

```python
# Now loop through that dictionary to print a readable summary
for courses, count in course_counts.items():
    print(f"{count} student(s) took {courses} course(s)")
```
