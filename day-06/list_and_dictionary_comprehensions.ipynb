# Day 06: List and Dictionary Comprehensions

**Dataset:** `Student_Marks_Dataset.csv`

This notebook uses list and dictionary comprehensions to create collections efficiently from
existing data, using real values pulled directly from the Student Marks dataset.

## 1. Import Libraries and Load the Dataset

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("../datasets/Student_Marks_Dataset.csv")

# Preview the data
df.head()
```

```python
# Basic info about the dataset
df.info()
```

**Pull the columns we'll use into plain Python lists**

```python
marks = list(df["Marks"])
number_courses = list(df["number_courses"])
time_study = list(df["time_study"])

print(marks[:10])
print(number_courses[:10])
```

## 2. List Comprehensions

The basic form is `[expression for item in iterable]` which it builds a new list by applying
`expression` to every item in `iterable`.

**Example 1: Round every mark to the nearest whole number**

```python
rounded_marks = [round(mark) for mark in marks]
print(rounded_marks[:10])
```

**Example 2: Keep only the passing marks (>= 40)**

```python
passing_marks = [mark for mark in marks if mark >= 40]
print(passing_marks[:10])
print("Number of passing marks:", len(passing_marks))
```

**Example 3: Round only the passing marks**

```python
rounded_passing_marks = [round(mark) for mark in marks if mark >= 40]
print(rounded_passing_marks[:10])
```

**Example 4: Label each mark Pass or Fail**

```python
pass_fail_labels = ["Pass" if mark >= 40 else "Fail" for mark in marks]
print(pass_fail_labels[:10])
```

**Example 5: Build a list of student labels**

```python
student_labels = [f"Student {i + 1}" for i in range(10)]
print(student_labels)
```

**Example 6: Comprehension over two lists at once, using `zip()`**

```python
courses_and_marks = [(courses, mark) for courses, mark in zip(number_courses[:10], marks[:10])]
print(courses_and_marks)
```

**Comparing a comprehension to a traditional loop**

Here's the same result for filtering passing marks written different ways to compare
readability and length.

```python
# Traditional for loop version
passing_marks_loop = []
for mark in marks:
    if mark >= 40:
        passing_marks_loop.append(mark)

# List comprehension version (yields identical result, written in one line)
passing_marks_comprehension = [mark for mark in marks if mark >= 40]

print(passing_marks_loop == passing_marks_comprehension)
print("Loop version needed 4 lines; comprehension version needed 1 line.")
```

## 3. Dictionary Comprehensions

The basic form is `{key_expression: value_expression for item in iterable}` which builds a new
dictionary in one line.

**Example 1: Map each student's position to their mark**

```python
marks_by_position = {i: marks[i] for i in range(len(marks[:10]))}
print(marks_by_position)
```

**Example 2: Keep only students who passed, with their mark**

```python
passing_students = {i: marks[i] for i in range(len(marks)) if marks[i] >= 40}
# Show just the first 10 entries for a readable preview
print(dict(list(passing_students.items())[:10]))
print("Number of passing students:", len(passing_students))
```

**Example 3: Pass/Fail status per student**

```python
pass_fail_by_position = {i: ("Pass" if marks[i] >= 40 else "Fail") for i in range(len(marks[:10]))}
print(pass_fail_by_position)
```

**Example 4: Count how many students took each number of courses**

```python
course_counts = {courses: number_courses.count(courses) for courses in set(number_courses)}
print(course_counts)
```

**Comparing a dictionary comprehension to a traditional loop**

Here's the same result for counting how many students took each number of courses written different ways to compare
readability and length.

```python
# Traditional for loop version
course_counts_loop = {}
for courses in set(number_courses):
    course_counts_loop[courses] = number_courses.count(courses)

# Dictionary comprehension version (identical result, written in one line)
course_counts_comprehension = {courses: number_courses.count(courses) for courses in set(number_courses)}

print(course_counts_loop == course_counts_comprehension)
print("Loop version needed 3 lines; comprehension version needed 1 line.")
```
