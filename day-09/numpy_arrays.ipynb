# Day 09: NumPy Array Creation and Data Types

**Dataset:** `Iris_Dataset.csv`

This notebook creates NumPy arrays and explores their shape, dimensions, size, data types,
and basic operations, using real values pulled directly from the Iris dataset.

## 1. Import Libraries and Load the Dataset

```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("../datasets/Iris_Dataset.csv")

# Preview the data
df.head()
```

```python
# Basic info about the dataset
df.info()
```

## 2. One-Dimensional Arrays

A 1D array is a single row (or column) of values, which is created with `np.array()`.

**Create a 1D array from a real column**

```python
sepal_length = np.array(df["sepal length (cm)"])
print(sepal_length[:10])
print(type(sepal_length))
```

**Create a 1D array manually from a plain Python list**

```python
manual_array = np.array([1, 2, 3, 4, 5])
print(manual_array)
print(type(manual_array))
```

## 3. Two-Dimensional Arrays

A 2D array has rows and columns, which useful for representing an entire table of features at
once.

**Create a 2D array from four real columns**

```python
feature_columns = ["sepal length (cm)", "sepal width (cm)", "petal length (cm)", "petal width (cm)"]
features = np.array(df[feature_columns])

print(features[:5])
print(type(features))
```

**Create a 2D array manually**

```python
manual_2d_array = np.array([[1, 2, 3], [4, 5, 6]])
print(manual_2d_array)
```

## 4. Array Attributes

Every NumPy array has attributes that describe its structure: `shape`, `ndim`, `size`, and
`dtype`.

**Attributes of the 1D array (`sepal_length`)**

```python
print("shape:", sepal_length.shape)   # (150,) - 150 values in one dimension
print("ndim:", sepal_length.ndim)      # 1 - one dimension
print("size:", sepal_length.size)      # 150 - total number of elements
print("dtype:", sepal_length.dtype)    # the data type of the elements
```

**Attributes of the 2D array (`features`)**

```python
print("shape:", features.shape)   # (150, 4) - 150 rows, 4 columns
print("ndim:", features.ndim)      # 2 - two dimensions
print("size:", features.size)      # 600 - total number of elements (150 x 4)
print("dtype:", features.dtype)
```

**Arrays can hold different data types — creating an array of strings**

```python
species = np.array(df["target"])
print(species[:5])
print("dtype:", species.dtype)   # a string dtype, not numeric
```

## 5. Numerical Operations on Arrays

**Summary statistics**

```python
print("Mean sepal length:", sepal_length.mean())
print("Max sepal length:", sepal_length.max())
print("Min sepal length:", sepal_length.min())
print("Sum of sepal length:", sepal_length.sum())
print("Standard deviation:", sepal_length.std())
```

**Element-wise arithmetic (every value is affected at once, with no loop needed)**

```python
sepal_length_doubled = sepal_length * 2
sepal_length_plus_one = sepal_length + 1

print(sepal_length[:5])
print(sepal_length_doubled[:5])
print(sepal_length_plus_one[:5])
```

**Operations between two arrays of the same shape**

```python
sepal_width = np.array(df["sepal width (cm)"])

# Adding two full columns together, element by element
sepal_total = sepal_length + sepal_width
print(sepal_total[:5])
```

**Boolean masking — filtering an array based on a condition**

```python
long_sepals = sepal_length[sepal_length > 6]
print(long_sepals[:10])
print("Number of flowers with sepal length > 6:", len(long_sepals))
```

**Statistics on the full 2D array, per column**

```python
column_means = features.mean(axis=0)
print("Mean of each column (sepal length, sepal width, petal length, petal width):")
print(column_means)
```

## 6. Comparing NumPy Arrays with Python Lists

The same numbers behave very differently depending on whether they're stored in a list or a
NumPy array.

```python
python_list = [1, 2, 3, 4, 5]
numpy_array = np.array([1, 2, 3, 4, 5])

# Multiplying a list by 2 repeats the whole list
print("List * 2:", python_list * 2)

# Multiplying a NumPy array by 2 multiplies every element
print("Array * 2:", numpy_array * 2)
```

```python
# Adding two lists together concatenates them, it does not add element-wise
python_list_2 = [10, 20, 30, 40, 50]
print("List + list:", python_list + python_list_2)

# Adding two arrays together adds them element-wise
numpy_array_2 = np.array([10, 20, 30, 40, 50])
print("Array + array:", numpy_array + numpy_array_2)
```

## Outcome

This notebook practiced creating one-dimensional and two-dimensional NumPy arrays using real
values from the `Iris_Dataset.csv` file, and explored their `shape`, `ndim`, `size`, and
`dtype` attributes. I performed numerical operations including summary statistics (`mean()`,
`max()`, `min()`, `sum()`, `std()`), element-wise arithmetic, operations between two arrays,
boolean masking to filter values by a condition, and per-column statistics on a 2D array using
`axis=0`. Comparing NumPy arrays directly against plain Python lists showed the key difference
that motivates using NumPy for data science: the same arithmetic operators (`*`, `+`) behave

foundation for fast, readable numerical computation on real data.
