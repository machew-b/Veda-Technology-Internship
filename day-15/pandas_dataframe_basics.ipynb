# Day 15: Pandas DataFrame Basics

**Dataset:** `Titanic_Dataset.csv`

This notebook creates DataFrames from a dictionary and from a real dataset, and explores
rows, columns, shape, and summary information, using real values pulled directly from the
Titanic dataset.

## 1. Import Libraries

```python
import pandas as pd
```

## 2. Create a DataFrame Manually

`pd.DataFrame()` can build a DataFrame directly from a Python dictionary, where each key
becomes a column name and each value (a list) becomes that column's data.

```python
manual_data = {
    "passenger_name": ["Alice Smith", "Bob Jones", "Carla Reyes"],
    "pclass": [1, 3, 2],
    "age": [29, 22, 35],
    "survived": [1, 0, 1]
}

manual_df = pd.DataFrame(manual_data)
manual_df
```

```python
print(type(manual_df))
print(manual_df.shape)
```

## 3. Load a Real Dataset into a DataFrame

```python
df = pd.read_csv("../datasets/Titanic_Dataset.csv")
df.head()
```

**A quick look at the last few rows, too**

```python
df.tail()
```

## 4. Display Shape, Columns, and Data Types

**`shape` — (number of rows, number of columns)**

```python
print(df.shape)
```

**`size` — total number of individual cells (rows × columns), for comparison**

```python
print(df.size)
print(df.shape[0] * df.shape[1], "should match df.size")
```

**`columns` — the column names**

```python
print(df.columns)
```

**`dtypes` — the data type of each column**

```python
print(df.dtypes)
```

**`info()` — a combined summary of shape, column names, non-null counts, and dtypes**

```python
df.info()
```

## 5. Summary Statistics

**`describe()` — summary statistics for every numeric column**

```python
df.describe()
```

**`describe(include="object")` — summary statistics for the text/categorical columns instead**

```python
df.describe(include="object")
```

**Checking for missing values in each column**

```python
df.isna().sum()
```

**A closer look at one categorical column: unique values and their counts**

```python
print(df["Embarked"].unique())
print(df["Embarked"].value_counts())
```

## Outcome

This notebook practiced building a Pandas DataFrame both manually (from a Python dictionary
using `pd.DataFrame()`) and from a real file (`Titanic_Dataset.csv` using `pd.read_csv()`). I
explored the loaded DataFrame's `shape`, `size`, `columns`, and `dtypes`, and used `info()` to
see all of that plus non-null counts in one combined summary — which immediately revealed real
missing data in the `Age`, `Cabin`, and `Embarked` columns. I used `describe()` to get summary
statistics (mean, min, max, quartiles) for the numeric columns, and `describe(include="object")`
to get a comparable summary (count, unique values, most frequent value) for the text columns.
This showed how a handful of Pandas methods can quickly answer the most important first
questions about any new dataset: how big is it, what columns does it have, what type is each
column, and where is data missing or unusual.
