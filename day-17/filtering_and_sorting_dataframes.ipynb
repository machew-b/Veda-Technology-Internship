# Day 17: Filtering and Sorting DataFrames

**Dataset:** `Superstore_Dataset.csv`

This notebook filters rows using conditions and sorts DataFrame values in ascending and
descending order, using real values pulled directly from the Superstore dataset.

## 1. Import Libraries and Load the Dataset

```python
import pandas as pd

# Load the dataset (encoding='latin1' avoids a UnicodeDecodeError on this file)
df = pd.read_csv("../datasets/Superstore_Dataset.csv", encoding="latin1")

# Preview the data
df.head()
```

```python
print(df.shape)
df.info()
```

## 2. Filtering Examples

Filtering uses a **boolean condition** inside `df[...]` (or `df.loc[...]`) to keep only the
rows where that condition is `True`.

**Example 1 — Orders with Sales greater than 500**

```python
high_sales = df[df["Sales"] > 500]
print(len(high_sales), "orders with Sales > 500")
high_sales[["Order ID", "Sales", "Category"]].head()
```

**Example 2 — Orders in the "Furniture" category**

```python
furniture_orders = df[df["Category"] == "Furniture"]
print(len(furniture_orders), "Furniture orders")
furniture_orders[["Order ID", "Category", "Sub-Category", "Sales"]].head()
```

**Example 3 — Loss-making orders (Profit less than 0)**

```python
loss_making = df[df["Profit"] < 0]
print(len(loss_making), "loss-making orders")
loss_making[["Order ID", "Category", "Profit"]].head()
```

**Example 4 — Combining multiple conditions with `&` (AND): loss-making Technology orders**

```python
loss_making_tech = df[(df["Category"] == "Technology") & (df["Profit"] < 0)]
print(len(loss_making_tech), "loss-making Technology orders")
loss_making_tech[["Order ID", "Sub-Category", "Sales", "Profit"]].head()
```

**Example 5 — Combining multiple conditions with `|` (OR): orders from the West or East region**

```python
west_or_east = df[(df["Region"] == "West") | (df["Region"] == "East")]
print(len(west_or_east), "orders from West or East")
west_or_east[["Order ID", "Region", "Sales"]].head()
```

**Example 6 — Filtering with `.isin()`: orders for specific sub-categories**

```python
selected_subcategories = df[df["Sub-Category"].isin(["Phones", "Chairs", "Tables"])]
print(len(selected_subcategories), "orders in Phones, Chairs, or Tables")
selected_subcategories[["Order ID", "Sub-Category", "Sales"]].head()
```

## 3. Sorting Examples

`sort_values()` reorders a DataFrame by one or more columns, in ascending order by default.

**Example 1 — Sort by Sales, descending (highest sales first)**

```python
df.sort_values("Sales", ascending=False)[["Order ID", "Sales"]].head(10)
```

**Example 2 — Sort by Profit, ascending (biggest losses first)**

```python
df.sort_values("Profit", ascending=True)[["Order ID", "Category", "Profit"]].head(10)
```

**Example 3 — Sort by multiple columns: Category (A-Z), then Sales (highest first) within each category**

```python
df.sort_values(["Category", "Sales"], ascending=[True, False])[["Category", "Sub-Category", "Sales"]].head(15)
```

**Example 4 — Sort the filtered loss-making orders by how large the loss is**

```python
loss_making.sort_values("Profit", ascending=True)[["Order ID", "Category", "Profit"]].head(10)
```

## 4. A Short Analysis of Filtered Records

Taking a closer look at the loss-making orders filtered above.

```python
print("Total loss-making orders:", len(loss_making))
print("Total profit lost:", round(loss_making["Profit"].sum(), 2))
print()
print("Loss-making orders by category:")
print(loss_making["Category"].value_counts())
print()
print("Average loss per order, by category:")
print(loss_making.groupby("Category")["Profit"].mean().round(2))
```

**Interpretation:** Out of 9,994 total orders, **1,871 (about 19%)** lost money, totaling
roughly **-$156,131** in lost profit. **Office Supplies** has the most loss-making orders
(886), followed by **Furniture** (714) and **Technology** (271) — but that's partly because
Office Supplies also has the most orders overall. Looking at the *average* loss per order by
category would be the next useful step to see which category loses the most money per
transaction, rather than just which has the most loss-making orders in total.

## Outcome

This notebook practiced filtering DataFrame rows using boolean conditions on the
`Superstore_Dataset.csv` file — single conditions (`Sales > 500`, `Category == "Furniture"`,
`Profit < 0`), combined conditions using `&` (AND) and `|` (OR), and membership filtering with
